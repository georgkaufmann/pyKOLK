# Read input parameter for `pyKOLK`

In this notebook, we introduce the functions **reading input data** for a solution pocket run.

----
## `readParameter2D()`

First, we read in the input file `KOLK_parameter.in`. 
We use the example from the `work` directory:

~~~
#=======================================================
#=== geometry ==========================================
#=======================================================
  0.40    801           # sidex [m], nx
  0.01                  # init_height [m]
#=======================================================
#=== define time series ================================
#=======================================================
     10                 # timeStep [a]
    500                 # timeWrite  [a]
#=======================================================
#=== climate ===========================================
#=======================================================
     8.      8.         # TSoilmin,TSoilmax [C]
   400.    400.         # PSoilmin,PSoilmax [ppm]
   400.    400.         # PAtmmin,PAtmmax [ppm]
     8.      8.         # TCavemin,TCavemax [C]
  1000.   1000.         # PCavemin,PCavemax [ppm]
    10.      1.         # dropCavemin,dropCavemax [s]
~~~

This file contains:
### geometry:
- `sidex`: [m] half width of model 
- `nx`: discretisation

### time series:

- `timeStep`: [a] time step of model
- `timeWrite`: [a] write time for model output

### climate: (will be used by time series for interpolation)

- `TSoilmin,TSoilmax`: [C] min/max values for soil temperature
- `PSoilmin,PSoilmax`: [ppm] min/max values for soil CO2
- `PAtmmin,PAtmmax`: [ppm] min/max values for atmospheric CO2
- `TCavemin,TCavemax`: [C] min/max values for cave temperature
- `PCavemin,PCavemax`: [ppm] min/max values for cave CO2
- `dropCavemin,dropCavemax`: [s] min/max values for cave drip interval

Next, we check the function info with `help`, then we run the function on example input.

```python
help(libKOLK.readParameter2D)
```

```python
libKOLK.readParameter2D(control=True)
```
