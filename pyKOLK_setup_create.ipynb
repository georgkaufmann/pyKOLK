# `pyKOLK`

pyKOLK, a program package prediction the shape of solution pockets in caves from pre-defined climate conditions.

```python
import numpy as np
import libKOLK
```

----
# Create grid for `pyKOLK`

In this notebook, we introduce the functions **creating the numerical grid** for a solution pocket run.

----
## `createGridD()`

This function creates the 2D grid of the current solution pocket surface.

It reads the geometry (`sidex`,`nx`) and the central height `init_height` of the initial surface.

It then creates the horizontal grid points `x`  as 1D array, 
and the vertical extension as grid points `y`, with a linear function from `init_height` in the
center towards zeros outside.

```python
help(libKOLK.createGrid2D)
```

```python
x,y,dx = libKOLK.createGrid2D(sidex=0.4,nx=21,init_height=0.1)
```

----
## `refineGridD()`

The function `refineGridD()` is used in insert a new point along the x-axis between points. 0 and 1.

```python
help(libKOLK.refineGrid2D)
```

```python
print('old grid\n',x,y)
x,y = libKOLK.refineGrid2D(x,y)
print('refined grid\n',x,y)
```
